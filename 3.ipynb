```python
import pandas as pd 
```

```python
data = pd.read_csv("dataset.csv")
data.head()
```

```python
data.shape
```

```python
statuses = [
    'Closed with explanation',
    'Closed with non-monetary relief',
    'Closed with monetary relief'
]

data = data[data['Company response to consumer'].isin(statuses).copy()]
```

```python
data['Meritorious'] = data['Company response to consumer'].map({'Closed with explanation': 0, 'Closed with non-monetary relief': 1, 'Closed with monetary relief': 1})
```

```python
data['Meritorious'].value_counts()
```

```python
data['Meritorious'].value_counts(normalize=True) * 100
```
